# Week 2 Assignment | Python and Temperature Conversion

This notebook asks for temperatures in Celsius, converts them to Fahrenheit, checks them against a threshold, uses a loop for several temperatures, and uses a function for the conversion.

```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32
```

```python
# Ask the user for one temperature and convert it
celsius = float(input("Enter a temperature in °C: "))
fahrenheit = celsius_to_fahrenheit(celsius)

print(f"{celsius:.1f} °C = {fahrenheit:.1f} °F")
```

```python
# Check whether the temperature is above or below a specified threshold
threshold = float(input("Enter the threshold in °F: "))

if fahrenheit > threshold:
    print("The temperature is above the threshold.")
elif fahrenheit < threshold:
    print("The temperature is below the threshold.")
else:
    print("The temperature is equal to the threshold.")
```

```python
# Use a loop to convert several temperatures
number_of_temperatures = int(input("How many temperatures would you like to convert? "))

for i in range(number_of_temperatures):
    temp_c = float(input(f"Enter temperature {i + 1} in °C: "))
    temp_f = celsius_to_fahrenheit(temp_c)
    print(f"{temp_c:.1f} °C = {temp_f:.1f} °F")
```
